```python
def add_numbers(a, b):
    return a + b

result = add_numbers(5, 10)
print("Sum:", result)
```

```python
def greet(name, message="Hello"):
    return f"{message}, {name}!"

greeting1 = greet("Alice")
greeting2 = greet("Bob", "Good morning")
print(greeting1)
print(greeting2)
```

```python
def describe_pet(pet_name, animal_type="dog"):
    print(f"I have a {animal_type} named {pet_name}.")

describe_pet(pet_name="Buddy", animal_type="cat")

describe_pet(animal_type="hamster", pet_name="Nibbles")
describe_pet(pet_name="Charlie")
```

```python
def make_pizza(size, *toppings):
    print(f"\n Making a {size}-inch pizza with the following toppings:")
    for topping in toppings:
        print(f"- {topping}")

make_pizza(12, "pepperoni", "mushrooms", "green peppers")
make_pizza(16, "extra cheese")
```

```python
def calculate(a, b):
    sum = a + b
    difference = a - b
    product = a * b
    quotient = a / b
    return sum, difference, product, quotient

s, d, p, q = calculate(10, 5)
print(f"Sum: {s}, Difference: {d}, Product: {p}, Quotient: {q}")

```

```python
x = 10

def example_function():
    x = 5
    print("Local x:", x)

example_function()
print("Global x:", x)
```

```python
x = 10

def modify_global():
    global x
    x = 20
    print("Modified global x:", x)

modify_global()
print("Global x after modification:", x)
```
