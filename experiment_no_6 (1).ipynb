```python
#SET
#PROGRAMME NO. 1
#EMPTY SET()
empty_set=set()
print(empty_set)
set1={1,"a",8.4,True,0}
print(set1)
```

```python
#PROGRAMME NO. 2
#SET METHODS
#ADD()
set1={1,2,3}
set1.add(10)
print(set1)


#remove()
set1={1,2,3}
set1.remove(1)
print(set1)
```

```python
#PROGRAMME NO. 3
#DISCARD()
set1={1,2,3}
set1.discard(5)
print(set1)

#POP()
set1={1,2,3}
popped_elements=set1.pop()
print(popped_elements)
print(set1)
```

```python
#PROGRAMME NO. 4
#UNION()
set1={1,2,3}
set2={4,5,6}
union_set=set1.union(set2)
union_set=set1|set2
print(union_set)
```

```python
#PROGRAMME NO. 5
#INTERSECTION
set1={1,2,5,8}
set2={3,4,7,8,2}
intersection_set=set1.intersection(set2)
intersection_set=set1-set2
print(union_set)
```

```python
#PEOGRAMME NO. 6
#DIFFERENCE
set1={"apple","banana","cherry"}
set2={"mango","orange","banana"}
difference_set=set1.difference(set2)
difference_set=set1-set2
print(difference_set)
```
