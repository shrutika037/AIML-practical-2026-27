```python
#conditional statement
#if else
age=25
if age>=18:
    print("eligible")
else:
    print("not eligible")
```

```python
#elseif(elif)
marks=370
if marks>=90:
     print("grade A")
elif marks>=75:
     print("grade B")
elif marks>=65:
     print("grade c")
else:
     print("grade D")
```

```python
#check whether no. is even or odd
a=int(input("enter the number :"))
if a%2==0:
    print("even")
else:
    print("odd")
```

```python
#nested if
age=12
citizen=True
if age>=18:
   if citizen:
      print("eligible")
else:
      print("not eligible")
```

```python
#using logical operators
#AND
age=20
if age>=18 and age<=60:
       print("can vote")
```

```python
#OR
age=25
if age>=23 or age<=20 :
         print("vote")
```

```python
#NESTED IF
age=20
if age>=18:
   if age<=60:
      print("drive")
   else:
      print("not drive")
```

```python
#NOT
is_holiday=False
if not is_holiday:
       print("go to office")
```

```python
#MULTIPLE CONDITIONS
salary=60000
experience=4
if salary>50000 and experience>3:
           print("get promotion")
else:
    print("not eligible for promotion")
```
